Make a bar plot with these three numbers. y is mean quarterly #paper published. x is categorical, corresponding to each bar's quarter range. 

```python
mean_by_range = pd.Series({
    "2014Q4-2022Q4": float(quarter_counts[quarter_counts.index < "2023Q1"].mean()),
    "2020Q1–2022Q4": float(quarter_counts[(quarter_counts.index >= "2020Q1") & (quarter_counts.index < "2023Q1")].mean()),
    "2023Q1+": float(quarter_counts[quarter_counts.index >= "2023Q1"].mean()),
})

ax = mean_by_range.plot(
    kind="bar",
    figsize=(8, 4),
    title="Mean Quarterly Papers by Quarter Range",
    xlabel="Quarter range",
    ylabel="Mean quarterly #paper published",
    rot=0,
    color=["red", "orange", "#003366"],  # deep blue
)

for i, v in enumerate(mean_by_range.values):
    ax.text(i, v, f"{v:.1f}", ha="center", va="bottom")
```
